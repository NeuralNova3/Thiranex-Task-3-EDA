```python

# ==========================================
# THIRANEX INTERNSHIP - TASK 3
# EXPLORATORY DATA ANALYSIS (EDA) PROJECT
# ==========================================

# Step 1: Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Step 2: Configure Pandas display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

# Step 3: Set visualization style
sns.set_style("whitegrid")

# Step 4: Display environment information
print("EDA Project Environment Ready!")
print("--------------------------------")
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
print("Matplotlib version:", plt.matplotlib.__version__)
print("Seaborn version:", sns.__version__)

print("--------------------------------")
print("All libraries imported successfully!")
```

```python
# Step 1: Load the roller coaster dataset

url = "https://gist.githubusercontent.com/Panchajanya1999/4310286fdb4e22f21c0726975f58b36e/raw/8f04fb5ef4ecc6ea37483b4aa73006a76a2d3449/coaster_db.csv"

df = pd.read_csv(url)

# Step 2: Display the first five rows
print("Dataset loaded successfully!")

display(df.head())
```

```python
# ==========================================
# CELL 3: DATASET OVERVIEW
# ==========================================

# 1. Check number of rows and columns
print("Dataset Shape:", df.shape)

# 2. Display all column names
print("\nColumn Names:")
print(df.columns.tolist())

# 3. Check dataset information
print("\nDataset Information:")
df.info()

# 4. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 5. Check duplicate records
print("\nDuplicate Rows:", df.duplicated().sum())

# 6. Generate statistical summary
print("\nStatistical Summary:")
display(df.describe(include="all").T)
```

```python
# ==========================================
# CELL 4: DATA CLEANING
# ==========================================

# Create a copy of the original dataset
df_clean = df.copy()

# 1. Standardize column names
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("/", "_")
)

# 2. Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# 3. Remove columns that are completely empty
df_clean = df_clean.dropna(axis=1, how="all")

# 4. Display cleaned dataset information
print("Data Cleaning Completed!")
print("--------------------------------")

print("Original Dataset Shape:", df.shape)
print("Cleaned Dataset Shape:", df_clean.shape)

print("\nMissing Values After Cleaning:")
print(df_clean.isnull().sum())

print("\nFirst Five Rows of Cleaned Dataset:")
display(df_clean.head())
```

```python
# ==========================================
# CELL 5: STATISTICAL ANALYSIS
# ==========================================

# Display statistical summary of numerical columns
print("STATISTICAL ANALYSIS")
print("====================")

display(df_clean.describe().T)

# Calculate average, maximum and minimum values
print("\nNumerical Data Analysis:")

numeric_df = df_clean.select_dtypes(include=np.number)

print("\nAverage Values:")
display(numeric_df.mean())

print("\nMaximum Values:")
display(numeric_df.max())

print("\nMinimum Values:")
display(numeric_df.min())
```

```python
# ==========================================
# CELL 6: DATA VISUALIZATION
# ==========================================

# 1. Distribution of roller coaster types
plt.figure(figsize=(10, 5))
df_clean['type'].value_counts().head(10).plot(kind='bar')
plt.title("Distribution of Roller Coaster Types")
plt.xlabel("Roller Coaster Type")
plt.ylabel("Number of Roller Coasters")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# 2. Distribution of roller coaster status
plt.figure(figsize=(8, 5))
df_clean['status'].value_counts().plot(kind='bar')
plt.title("Roller Coaster Status Distribution")
plt.xlabel("Status")
plt.ylabel("Number of Roller Coasters")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# 3. Distribution of opening years
plt.figure(figsize=(10, 5))
sns.histplot(df_clean['opening_date'].dropna(), bins=20, kde=True)
plt.title("Distribution of Roller Coaster Opening Dates")
plt.xlabel("Opening Date / Year")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()


# 4. Distribution of manufacturers
plt.figure(figsize=(10, 5))
df_clean['manufacturer'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Roller Coaster Manufacturers")
plt.xlabel("Manufacturer")
plt.ylabel("Number of Roller Coasters")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CELL 7: CORRELATION ANALYSIS
# ==========================================

# Select numerical columns
numeric_df = df_clean.select_dtypes(include=np.number)

# Calculate correlation
correlation = numeric_df.corr()

# Display correlation matrix
print("Correlation Matrix:")
display(correlation)

# Create correlation heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Roller Coaster Features")
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CELL 8: KEY FINDINGS AND INSIGHTS
# ==========================================

print("KEY FINDINGS FROM ROLLER COASTER DATA")
print("=" * 45)

# Total roller coasters
print("\n1. Total Roller Coasters:", len(df_clean))

# Most common roller coaster type
if 'type' in df_clean.columns:
    print("\n2. Most Common Type:")
    print(df_clean['type'].value_counts().head(1))

# Most common status
if 'status' in df_clean.columns:
    print("\n3. Roller Coaster Status:")
    print(df_clean['status'].value_counts().head())

# Top manufacturers
if 'manufacturer' in df_clean.columns:
    print("\n4. Top Manufacturers:")
    print(df_clean['manufacturer'].value_counts().head(5))

# Missing values
print("\n5. Missing Values:")
print(df_clean.isnull().sum().sum())

# Numerical observations
print("\n6. Numerical Data Summary:")
display(numeric_df.describe().T)

print("\nAnalysis completed successfully!")
```
